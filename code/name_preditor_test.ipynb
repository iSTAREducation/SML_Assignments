```python
import joblib
import pandas as pd

model_path = '../model/name_gender_predictor.pkl'
model = joblib.load(model_path)


name = "Myo Thida"  # Replace with the name you want to predict
y_pred = model.predict([name])[0]
gender = pd.Series(y_pred).map({0: 'male', 1: 'female'}).to_string().split()[1]

print("Predicted gender:", gender)
```
